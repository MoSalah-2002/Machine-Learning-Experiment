```python
def get_Q(states,actions,discount_factor,iter_num):
  prob_trans_states=env.P
  init_values={}
  for state in states:
    init_values[state]=0
  optimal_policy={}
  for state in states:
    optimal_policy[state]=0

  temp=[]
  res=0
  for iter in range(iter_num):
    # print(init_values)
    for state in states:
      temp=[0] * len(actions)
      for action in actions:
        for i in prob_trans_states[state][action]:
          p=i[0]
          r=i[2]
          res+=(p*(r+(discount_factor*init_values[i[1]])))
        temp[action]=res
        res=0
      init_values[state]=max(temp)
      optimal_policy[state] = np.argmax(temp)
  return optimal_policy



```

```python
final_moves=get_Q(range(state_num),range(action_state_num),0.9,1000)
final_moves2=list(final_moves.values())
```

```python
env.reset()
s=0
done=False
while not done:
    new_state,reward,done,info=env.step(final_moves2[s])
    s=new_state
    if done:
        break
```
