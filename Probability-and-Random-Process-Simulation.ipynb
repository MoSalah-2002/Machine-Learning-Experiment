```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
```

```python
days=365
currnet_day=365
prob=[]
num_of_persons=366
for i in range(num_of_persons):
  p=(currnet_day/days)
  if prob != []:
    p=p*prob[-1]
  prob.append(round(p,3))
  currnet_day-=1
plt.scatter(range(num_of_persons),prob)
plt.show()
```

```python
dice=np.arange(1,7)
num_of_toss=10000

res=[]
for i in range(num_of_toss):
  res.append(np.random.choice(dice))

counting=Counter(res)
nums=list(counting.keys())
prob=np.array(list(counting.values()))/num_of_toss
plt.bar(nums,prob)
plt.show()
```

```python
dice=np.arange(1,7)
prob=np.full(len(dice),1/len(dice))
prob=list(prob)
prob.insert(0,0)
```

```python
prob=np.array(prob)
CDF=prob.cumsum()
plt.bar(dice,CDF[1:],color="y")
plt.show()
```

```python
import sympy as sp

# Define the variable
x = sp.Symbol('x')

# Define the function
f = 1/(sp.sqrt(2*sp.pi)) * sp.gamma(-(x**2)/2)

# Indefinite integral
integral = sp.integrate(f, x)
result = sp.integrate(f, (x, -1.645, 1.645))
print(integral)
```
