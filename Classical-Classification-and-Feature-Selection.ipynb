```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
import plotly.express as px

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,Lasso,Ridge
from sklearn.svm import  SVR
from sklearn.ensemble import RandomForestRegressor
from sklearn.neighbors import KNeighborsRegressor


from sklearn.preprocessing import StandardScaler,LabelEncoder

from sklearn.metrics import mean_absolute_error,mean_squared_error

warnings.filterwarnings("ignore")
sns.set_style("darkgrid")
```

```python
diamonds=pd.read_csv("/content/diamonds.csv",index_col=0)
diamonds.drop(['depth',"table",'z'],axis=1,inplace=True)
diamonds.info()
```

```python
diamonds["Ratio"]=diamonds['x']/diamonds['y']
```

```python
diamonds.isnull().sum()
```

```python
diamonds.dropna(axis=0,inplace=True)
diamonds[diamonds['Ratio'].isnull()]
```

```python
from sklearn.preprocessing import LabelEncoder

LE_cut = LabelEncoder()
cut2 = LE_cut.fit_transform(diamonds['cut'])
diamonds['cut2'] = cut2

LE_color = LabelEncoder()
color2 = LE_color.fit_transform(diamonds['color'])
diamonds['color2'] = color2

LE_clarity = LabelEncoder()
clarity2 = LE_clarity.fit_transform(diamonds['clarity'])
diamonds['clarity2'] = clarity2
```

```python

# Get class mapping
print("Cut mapping:")
for cls, num in zip(LE_cut.classes_, range(len(LE_cut.classes_))):
    print(f"{cls} -> {num}")

print("\nColor mapping:")
for cls, num in zip(LE_color.classes_, range(len(LE_color.classes_))):
    print(f"{cls} -> {num}")

print("\nClarity mapping:")
for cls, num in zip(LE_clarity.classes_, range(len(LE_clarity.classes_))):
    print(f"{cls} -> {num}")
```

```python
diamonds.head()
```

```python
X=diamonds.drop(["cut","color","clarity","x","y","price"],axis=1)
Y=diamonds[["price"]]
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2,shuffle=True,random_state=255)
```

```python
x_train
```

```python
model=RandomForestRegressor()
model.fit(x_train,y_train)
y_p=model.predict(x_test)
y_p[:10],y_test[:10]
```

```python
mean_absolute_error(y_test,y_p)
```
